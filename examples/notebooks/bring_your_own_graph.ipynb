# Bring your own graph to CascadeLens
This notebook imports a CSV dependency graph and runs bounded scenario analysis. Imported topology remains `MODEL_INFERRED`; the result is not an empirical forecast.

```python
from pathlib import Path
from cascadelens import analyze, default_scenario, graph_from_csv
```

```python
csv_text = Path('../data/simple_edges.csv').read_text(encoding='utf-8')
snapshot = graph_from_csv(csv_text, decision_cutoff='2026-01-01T00:00:00Z')
scenario = default_scenario(snapshot, magnitude=0.6)
result = analyze(snapshot, scenario)
```

```python
[
    {'bound': bound, 'impact': result.bounds[bound]['totalWeightedImpact']}
    for bound in ('lower', 'central', 'upper')
]
```

The lower and central bounds exclude imported `MODEL_INFERRED` dependency edges. The upper bound includes them. This gap is an evidence boundary, not a confidence interval.
